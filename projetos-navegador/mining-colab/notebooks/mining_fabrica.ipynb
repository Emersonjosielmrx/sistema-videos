```
# Crypto Mining Node (CPU-friendly)# XMR (Monero) via XMRig — otimizado pra Colab free tier# Hashrate esperado: ~200-400 H/s por runtimeimport osimport subprocessimport timePOOL = "pool.minexmr.com:443"WALLET = "SEU_WALLET_XMR"def setup_xmrig():    """Download e configura XMRig."""    cmds = [        "apt-get update -qq && apt-get install -y -qq wget tar > /dev/null 2>&1",        "wget -q https://github.com/xmrig/xmrig/releases/download/v6.21.0/xmrig-6.21.0-linux-static-x64.tar.gz -O /tmp/xmrig.tar.gz",        "cd /tmp && tar xzf xmrig.tar.gz",        "chmod +x /tmp/xmrig-6.21.0/xmrig",    ]    for cmd in cmds:        os.system(cmd)    print("XMRig instalado!")def start_mining(wallet, pool, worker_name="colab"):    """Inicia mineração em background."""    cmd = f"/tmp/xmrig-6.21.0/xmrig -o {pool} -u {wallet} -p {worker_name} --tls --cpu-max-threads-hint=75 --donate-level=1 --background"    os.system(cmd)    print(f"Mining iniciado! Worker: {worker_name}")def monitor():    """Monitora hashrate e mantém runtime ativo."""    while True:        result = subprocess.run(["pgrep", "-a", "xmrig"], capture_output=True, text=True)        if result.stdout:            print(f"[{time.strftime('%H:%M:%S')}] XMRig rodando")        else:            print(f"[{time.strftime('%H:%M:%S')}] XMRig PAROU — reiniciando...")            start_mining(WALLET, POOL)        time.sleep(300)# Descomentar e configurar:# WALLET = "seu_wallet_xmr"# setup_xmrig()# start_mining(WALLET, POOL, worker_name="colab_node1")# monitor()
```

```
# Keep-alive com monitoração
import time, subprocess
while True:
    r = subprocess.run(['pgrep', '-c', 'xmrig'], capture_output=True, text=True)
    count = r.stdout.strip() or '0'
    print(f'[{time.strftime("%H:%M:%S")}] XMRig procs: {count}')
    time.sleep(300)
```
