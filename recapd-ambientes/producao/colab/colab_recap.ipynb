```
import os, shutil, site, subprocess, sys
from pathlib import Path
C = '/content/drive/MyDrive/colab_recap'
def rodar(cmd):
    r = subprocess.run(cmd, capture_output=True, text=True)
    print(r.stdout[-3000:], r.stderr[-3000:])
    assert r.returncode == 0, cmd
rodar([sys.executable, '-m', 'pip', 'install', '-q', 'edge-tts==7.3.1', 'opencv-python-headless', 'psutil', 'pydub', 'soundfile'])
rodar([sys.executable, f'{C}/baixar_drive.py'])
sp = Path(site.getsitepackages()[0])
shutil.copy(f'{C}/sitecustomize.py', sp / 'colab_traduz.py')
(sp / 'colab_traduz.pth').write_text('import colab_traduz\n')
prod = Path('/content/raiz/D/recapd-ambientes/producao')
(prod / 'workspace' / 'anime').mkdir(parents=True, exist_ok=True)
if not (prod / 'workspace' / 'assets').exists():
    os.symlink('/content/raiz/D/hipercool-materiais/assets', prod / 'workspace' / 'assets')
(prod / 'venv' / 'Scripts').mkdir(parents=True, exist_ok=True)
py = prod / 'venv' / 'Scripts' / 'python.exe'
if not py.exists():
    os.symlink(sys.executable, py)
print('pronto')
```
